```python
import pandas as pd
#设置展示行数
pd.set_option('display.min_rows', 30)
#读入文件
df1=pd.read_csv("csv_data/订单数据_待清洗.csv")
#缺失值处理
print(df1.isnull())
#删除缺失值
df=df1.dropna()#删除缺失值所在行
df=df1.dropna(axis=1)#删除缺失值所在列 需要用值去接收
#填充缺失值
# df1.fillna("--")
# df1.ffill() #用上一行数据填充
# df1.bfill() #用下一行数据填充
```

```python
#重复值的处理
#查看重复值
df1.duplicated() #所有列都重复 才会
df1.duplicated(subset=["订单号"])
#删除重复值
df1.drop_duplicates(subset=["订单号"], keep="first")#默认保留第一个重复值
# last 保留最后一个 False 两个都不保留
```

```python
#异常值的处理
#查看异常值
df1[df1["单价"]<0]
#删除异常值
df1.drop(df1[df1["单价"]<0].index)#获得异常值的索引
#修复异常值
df1["单价"]=df1["单价"].abs()#取绝对值
```

```python
#数据格式转换
df1["订单日期"]=df1["订单日期"].replace("2025/06/05","2025-06-05")
df1["订单日期"]=df1["订单日期"].str.replace()#内部有转换器 可以直接调用为字符串进行操作
```
