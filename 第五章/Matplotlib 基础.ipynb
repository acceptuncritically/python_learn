```python
import matplotlib.pyplot as plt
import random



plt.rcParams['font.sans-serif'] = ['SimHei'] #设置中文字体为黑体
#目标 绘制折线图
x=[i for i in range(1,25)]
# y=[5,7,9,20,10,12,23,34,45,52,63,73,85]
y_beijing=[random.randint(10,18) for i in x]
y_xian=[random.randint(14,20) for i in x]
#设置自己的画布
plt.figure(figsize=(10,5)) #figsize设置画布大小
plt.plot(x,y_beijing,label="北京")#如果没有会自己创建
plt.plot(x,y_xian,label="西安")
#设置标题信息
plt.title("气温变化折线图",fontsize=20) #fontsize 设置标题的大小
plt.xlabel("时间")
# plt.xticks(x[1::2])
plt.xticks(x)
plt.ylabel("温度")
plt.yticks([i for i in range(5,25)])
plt.grid(linestyle="--",alpha=0.8)#生成网格线
# plt.legend(["北京","西安"])#显示图例
plt.legend(loc="best")#显示图例
plt.show()
```
