```python
#数据选择
df2=pd.read_csv("csv_data/学生成绩.csv",usecols=["姓名","语文","数学","英语"],index_col="数学")
#index_col 可以自己选择索引为那一列
# 单列
df2["姓名"]
df2[["语文","英语"]]#在指定索引后就不能继续调用了
df2.iloc[0:5]#行号一开始就有 从零开始
```

```python
# 数据的过滤
df3=pd.read_csv("csv_data/学生信息.csv")
pd.set_option("display.max_rows",None)
#单条件
df3[df3["年龄"]>=21]
df3[df3["班级"].isin(["食品","图书"])]
df3[df3["年龄"].between(20,21)]#between默认包含边界还有“neither", "left", "right"
#双条件 需要将小括号括起来
df3[(df3["性别"]=="女")&(df3["年龄"]>21)]

print(df3[(df3["性别"].isin(["男","女"]))&(df3["班级"].isin(["计算机1班","软件1班"]))])
```
