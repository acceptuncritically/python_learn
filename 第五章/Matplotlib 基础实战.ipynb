```python
from matplotlib.axes import Axes
#创建子图
import matplotlib.pyplot as plt
# figure,axes=plt.subplots(nrows=2,ncols=2)
# 用axes[0,0] 形式表示 先是第几行 再是第几列 axes[0,1]表示右上角的图
plt.rcParams['font.sans-serif'] = ['SimHei']
# figure 是画布对象 axes 是子图数组
figure,axes=plt.subplots(nrows=1, ncols=2, figsize=(20,6),dpi=100) #dpi 表示分辨率
#图一绘制柱状图
countrys=["中国","美国","沙特","伊朗","加拿大","印度"]
values=[35,25,16,73,21,78]
axes1:Axes=axes[0]
axes1.bar(countrys,values,color="darkblue",width=0.5)
axes1.set_title("世界石油储备",fontsize=15)
axes1.set_xlabel("国家")
axes1.set_ylabel("亿吨")
axes1.grid(linestyle="--",alpha=0.3)

#图二 饼状图 世界人口
# 国家列表
countries = ["印度", "中国", "美国", "印度尼西亚", "巴基斯坦", "尼日利亚", "巴西","其他"]

# 对应人口列表（单位：亿人）
populations = [14.28, 14.25, 3.40, 2.78, 2.40, 2.23, 2.16,38.5]
axes2:Axes=axes[1]
axes2.pie(populations,labels=countries,autopct="%1.1f%%")#左右的%表示里面的%原样输出，1.1 前的1表示不足位数会用空格补足位数 后面的1 表示保留小数点后一位
axes2.set_title("世界人口比例")
axes2.legend(loc="best")#bbbox_to_anchor(左右偏移量，上下偏移量)
#保存图片
plt.savefig("csv_data/图标.png")
plt.show()

```
